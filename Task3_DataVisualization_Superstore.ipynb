```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Sample - Superstore.csv",encoding="latin1")
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x="Category",y="Sales",data=df,estimator=sum,errorbar=None)
plt.title("Total Sales by Category")
plt.ylabel("Total Sales")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x="Sub-Category",y="Profit",data=df, estimator=sum, errorbar=None)
plt.title("Total Profit by Sub-Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.show()
```

```python
df["Order Date"]= pd.to_datetime(df["Order Date"])
monthly_sales=df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()
plt.figure(figsize=(12,6))
monthly_sales.plot(kind="line",marker="o")
plt.title("Monthly Sales Trend")
plt.ylabel("Total Sales")
plt.xlabel("Month")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x="Region",y="Sales",data=df,estimator=sum,errorbar=None)
plt.title("Total Sales by Regon")
plt.ylabel("Total Sales")
plt.show()
```

```python
fig, axes = plt.subplots(2,2,figsize=(16,10))
sns.barplot(x="Category",y="Sales",data=df,estimator=sum,errorbar=None,ax=axes[0,0])
axes[0,0].set_title("Total Sales by Category")

sns.barplot(x="Sub-Category",y="Profit",data=df, estimator=sum,errorbar=None,ax=axes[0,1])
axes[0,1].set_title("Total Profit by Sub-Category")
axes[0,1].tick_params(axis="x",rotation=45)

monthly_sales.plot(kind="line",marker="o",ax=axes[1,0])
axes[1,0].set_title("Total Sales by Trend")
axes[1,0].tick_params(axis="x",rotation=45)

sns.barplot(x="Region",y="Sales",data=df,estimator=sum,errorbar=None,ax=axes[1,1])
axes[1,1].set_title("Total Profit by Region")

plt.tight_layout()
plt.show
```

Technology is the selling category,followed closely by furniture and office Supplies.
Tables and Bookcases are loss making despite sales - a pricing or discout issue likely needs review.
Copies ,phones,chairs are the strongest profit drivers.
The west and east regions generate the highest sales,while south logs behind_ a potential area for targetted marketing.
