```python
import pandas as pd
df=pd.read_csv("Sample - Superstore.csv",encoding="latin1")
df.head()
```

```python
print(df.shape)
df.info()
df.describe()
```

4.How does discoubnt level relate to profit?

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
print(category_sales)

category_profit = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)
print(category_profit)
```

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
print(region_sales)

region_profit = df.groupby("Region")["Profit"].sum().sort_values(ascending=False)
print(region_profit)
```

```python
loss_orders = df[df["Profit"]<0]
print("Number of loss-making orders:",loss_orders.shape[0])

print(loss_orders["Category"].value_counts())
print(loss_orders["Sub-Category"].value_counts())
print(loss_orders["Discount"].describe())
```

```python
correlation = df["Discount"].corr(df["Profit"])
print("Correlation between Discount and Profit:",correlation)
df.groupby("Discount")["Profit"].mean().sort_index()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Order Month"] = df["Order Date"].dt.to_period("M")

monthly_sales = df.groupby("Order Month")["Sales"].sum()
print(monthly_sales)
```

*Dataset overview:9994 orders 21 columns,no missing values,covering Sales/Profit/Discount across Category,Region and time.  

*Data isue found: 834 orders at 8% are sold at a loss, concentrated in office supplies_especiallybinders.

*Seasonal trend: Sales consistently dip in jan/feb and peak in sep,nov,and dec each year - likely holiday-driven demand.
